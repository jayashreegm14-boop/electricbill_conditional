```python
#electric bill calculation system
consumer_name = input("enter your name:")
consumer_number = input("enter your mobile number:")
consumer_id = input("enter your consumer id:")
traiff = input("enter your traiff(commercial or domestic):").strip().lower()
unit_price = int(input("enter your used unit:"))
subsidy = 0
new_subsidy = 369
Tax = 0

#using conditional statement
if traiff == "domestic":
    if unit_price <= 200:
        Total = (unit_price - 200) * 2
        subsidy = 0
    elif unit_price <= 300:
        Total = (unit_price - 200) * 3
        subsidy = 30
    else:
        Total = (unit_price - 200) * 9
        subsidy = 90

elif traiff == "commercial":
    if unit_price <= 200:
        Total = (unit_price - 200) * 10
        Tax = 150
    elif unit_price <= 500:
        Total = (unit_price - 200) * 14
        Tax = 250
    else:
        Total = (unit_price - 200) * 15
        Tax = 300

else:
    Total = (unit_price - 200) * 20
    Tax = 450

#print display the input
print("=====WELCOME TO POWER TRANSMISSION GRID BILL OFFICE=====")
print("\nconsumer name:", consumer_name,
      "\nconsumer number:", consumer_number,
      "\nconsumer id:", consumer_id,
      "\ntraiff:", traiff,
      "\nsubsidy:", subsidy,
      "\nnew subsidy:", new_subsidy)

if traiff == "domestic":
    final_bill = Total + subsidy - new_subsidy
    print("final amount:", final_bill)
else:
    final_bill = Total + Tax
    print("final amount:", final_bill)
```
